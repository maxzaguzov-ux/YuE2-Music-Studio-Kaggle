```python
# ============================================================
# 4. YuE 2.0 Studio Master (Interactive Music Generation with ipywidgets)
# ============================================================
import ipywidgets as widgets
from IPython.display import display, HTML
import json
import time
import random
import glob
import os
import uuid
import subprocess
import websocket
import urllib.request
import base64

# Preset styles
preset_dict = {
    "Indie Pop Anthem": "Upbeat indie pop with warm female vocals, bright electric guitars, punchy drums, melodic bass, and subtle synth layers. Catchy and energetic, with an uplifting summer atmosphere, a memorable chorus, and polished modern production.",
    "Synthwave 80s Retro": "1980s retro synthwave, analog vintage synthesizers, gated snare reverb, driving arpeggiated bassline, male nostalgic vocals, neon aesthetic, cinematic nocturnal drive.",
    "Lo-Fi Chill Hop": "Warm lofi hip hop, vinyl crackle, mellow electric piano chords, smooth relaxed boom bap drums, deep warm sub bass, peaceful afternoon coffee shop atmosphere.",
    "Cinematic Epic Orchestral": "Epic cinematic trailer music, soaring brass ensemble, thundering Taiko drums, expressive emotional strings, dramatic crescendos, heroic orchestral climax.",
    "Modern Alt Rock": "Modern alternative rock, overdriven crunchy guitars, aggressive live acoustic drum kit, energetic lead vocals, heavy bassline, anthemic chorus."
}

# Create widgets
genre_dropdown = widgets.Dropdown(
    options=["Custom", "Indie Pop Anthem", "Synthwave 80s Retro", "Lo-Fi Chill Hop", "Cinematic Epic Orchestral", "Modern Alt Rock"],
    value="Indie Pop Anthem",
    description='Genre:',
    style={'description_width': 'initial'}
)

style_textarea = widgets.Textarea(
    value=preset_dict["Indie Pop Anthem"],
    description='Style:',
    layout=widgets.Layout(width='100%', height='100px'),
    style={'description_width': 'initial'}
)

lyrics_textarea = widgets.Textarea(
    value="""[Verse]
Morning light across the window
City waking down below
I can hear the streets are calling
Feels like somewhere we should go

[Chorus]
Run with me into the sunlight
Leave the shadows far behind
We don't need to know tomorrow
Tonight the whole world feels alive

[Verse]
Radio playing through the open door
Laughing like we did before
Every mile becomes a memory
And I just want a little more

[Chorus]
Run with me into the sunlight
Leave the shadows far behind
We don't need to know tomorrow
Tonight the whole world feels alive

[Outro]""",
    description='Lyrics:',
    layout=widgets.Layout(width='100%', height='200px'),
    style={'description_width': 'initial'}
)

abc_checkbox = widgets.Checkbox(
    value=False,
    description='Enable ABC Planning (slower, +2-3 min)',
    style={'description_width': 'initial'}
)

composition_radio = widgets.RadioButtons(
    options=['full', 'melody'],
    value='full',
    description='Mode:',
    style={'description_width': 'initial'}
)

abc_tokens_dropdown = widgets.Dropdown(
    options=[1024, 2048, 4096, 8192],
    value=8192,
    description='ABC Tokens:',
    style={'description_width': 'initial'}
)

sampler_dropdown = widgets.Dropdown(
    options=["dpm_2", "euler", "dpmpp_2m", "heun", "dpmpp_sde"],
    value="dpm_2",
    description='Sampler:',
    style={'description_width': 'initial'}
)

scheduler_dropdown = widgets.Dropdown(
    options=["sgm_uniform", "normal", "karras", "simple"],
    value="sgm_uniform",
    description='Scheduler:',
    style={'description_width': 'initial'}
)

steps_slider = widgets.IntSlider(
    value=32,
    min=10,
    max=50,
    step=1,
    description='Steps:',
    style={'description_width': 'initial'}
)

randomize_checkbox = widgets.Checkbox(
    value=True,
    description='Randomize Seed',
    style={'description_width': 'initial'}
)

seed_input = widgets.IntText(
    value=42,
    description='Seed:',
    disabled=True,
    style={'description_width': 'initial'}
)

format_radio = widgets.RadioButtons(
    options=['mp3', 'wav', 'flac'],
    value='mp3',
    description='Format:',
    style={'description_width': 'initial'}
)

quality_dropdown = widgets.Dropdown(
    options=["320k", "256k", "192k", "128k"],
    value="320k",
    description='Quality:',
    style={'description_width': 'initial'}
)

generate_button = widgets.Button(
    description='🎵 Generate Music',
    button_style='primary',
    layout=widgets.Layout(width='100%', height='50px')
)

output_area = widgets.Output()

# Update style when genre changes
def update_style(change):
    if change['new'] != "Custom":
        style_textarea.value = preset_dict.get(change['new'], "")

genre_dropdown.observe(update_style, names='value')

# Toggle seed input
def toggle_seed(change):
    seed_input.disabled = change['new']

randomize_checkbox.observe(toggle_seed, names='value')

# Generate function
def on_generate_clicked(b):
    with output_area:
        output_area.clear_output()
        
        # Get values from widgets
        genre_preset = genre_dropdown.value
        custom_style = style_textarea.value
        lyrics = lyrics_textarea.value
        enable_abc = abc_checkbox.value
        composition_mode = composition_radio.value
        max_abc_tokens = abc_tokens_dropdown.value
        sampler_name = sampler_dropdown.value
        scheduler = scheduler_dropdown.value
        steps = steps_slider.value
        randomize_seed = randomize_checkbox.value
        seed = seed_input.value
        audio_format = format_radio.value
        audio_quality = quality_dropdown.value
        
        if not lyrics or not lyrics.strip():
            print("❌ Please enter lyrics!")
            return
        
        display(HTML(f"""
        <div style='background-color:#1e1e2f;color:#fff;padding:16px;border-radius:10px;margin-bottom:15px;'>
          <h3 style='margin:0 0 10px 0;color:#ff79c6;'>🚀 YuE2 Job Dispatched</h3>
          <div style='display:grid;grid-template-columns:repeat(auto-fit,minmax(180px,1fr));gap:10px;font-size:13px;'>
            <div><b>Genre:</b> <span style='color:#50fa7b;'>{genre_preset}</span></div>
            <div><b>ABC Planning:</b> <span style='color:#f1fa8c;'>{'Enabled' if enable_abc else 'Bypassed'}</span></div>
            <div><b>Sampler:</b> <span style='color:#bd93f9;'>{sampler_name} ({steps} steps)</span></div>
            <div><b>Format:</b> <span style='color:#8be9fd;'>{audio_format} {audio_quality}</span></div>
          </div>
        </div>
        """))
        
        # Prepare parameters
        active_style = custom_style if genre_preset == "Custom" else preset_dict.get(genre_preset, custom_style)
        active_seed = random.randint(1, 2147483647) if randomize_seed else seed
        ckpt_name = "yue2_3b_fp16.safetensors"
        MAX_DURATION_SECONDS = 360
        
        workflow_api = {
          "10": {
            "inputs": {
              "filename_prefix": f"audio/YuE2_Widget_{int(time.time())}",
              "format": audio_format,
              "format.quality": audio_quality,
              "audio": ["33:9", 0]
            },
            "class_type": "SaveAudioAdvanced"
          },
          "33:15": {
            "inputs": {"ckpt_name": ckpt_name},
            "class_type": "CheckpointLoaderSimple"
          },
          "33:24": {
            "inputs": {
              "style": ["33:36", 0],
              "lyrics": ["33:37", 0],
              "seed": ["33:34", 0],
              "mode": composition_mode,
              "max_abc_tokens": max_abc_tokens,
              "temperature": 0.7,
              "top_p": 0.9,
              "top_k": 30,
              "repetition_penalty": 1.005,
              "penalty_window": 100,
              "clip": ["33:15", 1]
            },
            "class_type": "YuE2GenerateABC"
          },
          "33:14": {
            "inputs": {"source": ["33:30", 0]},
            "class_type": "PreviewAny"
          },
          "33:30": {
            "inputs": {
              "switch": ["33:31", 0],
              "on_false": ["33:32", 0],
              "on_true": ["33:24", 0]
            },
            "class_type": "ComfySwitchNode"
          },
          "33:32": {
            "inputs": {"value": ""},
            "class_type": "PrimitiveString"
          },
          "33:25": {
            "inputs": {
              "style": ["33:36", 0],
              "lyrics": ["33:37", 0],
              "abc": ["33:14", 0],
              "seed": ["33:34", 0],
              "mode": composition_mode,
              "max_duration": MAX_DURATION_SECONDS,
              "temperature": 1.0,
              "top_p": 0.95,
              "top_k": 100,
              "repetition_penalty": 1.2,
              "clip": ["33:15", 1]
            },
            "class_type": "YuE2GenerateMusic"
          },
          "33:31": {
            "inputs": {"value": enable_abc},
            "class_type": "PrimitiveBoolean"
          },
          "33:18": {
            "inputs": {"conditioning": ["33:25", 0]},
            "class_type": "ConditioningZeroOut"
          },
          "33:23": {
            "inputs": {"source": ["33:25", 1]},
            "class_type": "PreviewAny"
          },
          "33:8": {
            "inputs": {
              "seed": ["33:34", 0],
              "steps": steps,
              "cfg": 1.0,
              "sampler_name": sampler_name,
              "scheduler": scheduler,
              "denoise": 1.0,
              "model": ["33:15", 0],
              "positive": ["33:25", 0],
              "negative": ["33:18", 0],
              "latent_image": ["33:5", 0]
            },
            "class_type": "KSampler"
          },
          "33:5": {
            "inputs": {
              "seconds": ["33:25", 1],
              "batch_size": 1
            },
            "class_type": "EmptyYuE2LatentAudio"
          },
          "33:9": {
            "inputs": {
              "samples": ["33:8", 0],
              "vae": ["33:15", 2]
            },
            "class_type": "VAEDecodeAudio"
          },
          "33:34": {
            "inputs": {"seed": active_seed},
            "class_type": "SeedNode"
          },
          "33:36": {
            "inputs": {"value": active_style},
            "class_type": "PrimitiveStringMultiline"
          },
          "33:37": {
            "inputs": {"value": lyrics},
            "class_type": "PrimitiveStringMultiline"
          }
        }
        
        node_names = {
            "33:15": f"CheckpointLoaderSimple ({ckpt_name})",
            "33:24": "YuE2GenerateABC (Symbolic Score Planner)",
            "33:30": "ComfySwitchNode (ABC Bypass Route)",
            "33:25": "YuE2GenerateMusic (Acoustic Audio Synthesis)",
            "33:8": f"KSampler (Diffusion Decoding - {steps} steps)",
            "33:9": "VAEDecodeAudio (Audio Decompression)",
            "10": "SaveAudioAdvanced (Export)"
        }
        
        try:
            print("📡 Connecting to ComfyUI...")
            client_id = str(uuid.uuid4())
            ws = websocket.WebSocket()
            ws.connect(f"ws://127.0.0.1:8188/ws?clientId={client_id}")
            
            req_payload = json.dumps({"prompt": workflow_api, "client_id": client_id}).encode('utf-8')
            req = urllib.request.Request("http://127.0.0.1:8188/prompt", data=req_payload, headers={'Content-Type': 'application/json'})
            resp = urllib.request.urlopen(req)
            prompt_id = json.loads(resp.read())['prompt_id']
            
            print(f"📡 [Prompt ID: {prompt_id}] Active Seed: {active_seed}")
            print("=" * 65)
            
            start_time = time.time()
            job_failed = False
            
            while True:
                msg = ws.recv()
                if isinstance(msg, str):
                    event = json.loads(msg)
                    event_type = event.get('type')
                    event_data = event.get('data', {})
                    
                    if event_type == 'execution_error':
                        print(f"\n❌ Pipeline Error: {event_data.get('exception_message')}")
                        job_failed = True
                        break
                    
                    if event_type == 'executing':
                        node_id = event_data.get('node')
                        if node_id is None:
                            if event_data.get('prompt_id') == prompt_id:
                                print("\n\n🎉 Pipeline Completed!")
                                break
                        else:
                            stage_label = node_names.get(str(node_id), f"Node {node_id}")
                            print(f"\n▶️ [Stage: {stage_label}]")
                    
                    elif event_type == 'progress':
                        curr_val = event_data.get('value', 0)
                        max_val = event_data.get('max', steps)
                        pct = int((curr_val / max_val) * 100)
                        bar_len = 25
                        filled = int(bar_len * curr_val // max_val)
                        bar = '█' * filled + '░' * (bar_len - filled)
                        elapsed_s = int(time.time() - start_time)
                        print(f"\r   Sampling: [{bar}] {curr_val}/{max_val} ({pct}%) | {elapsed_s}s elapsed", end="", flush=True)
            
            ws.close()
            total_seconds = round(time.time() - start_time, 1)
            print(f"\n⏱️ Total time: {total_seconds}s")
            
            if not job_failed:
                candidate_files = glob.glob(f"/kaggle/working/ComfyUI/output/audio/YuE2_Widget_*.{audio_format}")
                
                if candidate_files:
                    latest_file = max(candidate_files, key=os.path.getctime)
                    file_size_mb = round(os.path.getsize(latest_file) / (1024 * 1024), 2)
                    filename_base = os.path.basename(latest_file)
                    
                    try:
                        _d = float(subprocess.check_output(
                            ["ffprobe","-v","error","-show_entries","format=duration","-of","default=nw=1:nk=1", latest_file]).strip())
                        actual_duration_txt = f"{int(_d//60)}:{int(_d%60):02d}"
                    except Exception:
                        actual_duration_txt = "n/a"
                    
                    with open(latest_file, "rb") as f:
                        audio_b64 = base64.b64encode(f.read()).decode("utf-8")
                    
                    mime_type = "audio/mpeg" if audio_format == "mp3" else f"audio/{audio_format}"
                    
                    display(HTML(f"""
                    <div style='background:#121214;border:1px solid #27272a;border-radius:14px;padding:22px;max-width:680px;font-family:sans-serif;color:#f4f4f5;box-shadow:0 10px 25px rgba(0,0,0,0.5);margin-top:15px;'>
                      <div style='display:flex;align-items:center;justify-content:space-between;margin-bottom:14px;'>
                        <div style='display:flex;align-items:center;gap:10px;'>
                          <div style='background:linear-gradient(135deg,#ec4899,#8b5cf6);width:42px;height:42px;border-radius:10px;display:flex;align-items:center;justify-content:center;font-size:22px;'>🎵</div>
                          <div>
                            <h3 style='margin:0;font-size:16px;color:#fff;'>{filename_base}</h3>
                            <span style='color:#a1a1aa;font-size:12px;'>Rendered in <b>{total_seconds}s</b> • {file_size_mb} MB • {actual_duration_txt} Audio</span>
                          </div>
                        </div>
                        <span style='background:#22c55e20;color:#4ade80;padding:4px 10px;border-radius:20px;font-size:11px;font-weight:bold;'>READY</span>
                      </div>
        
                      <audio controls autoplay style='width:100%;margin:12px 0;border-radius:8px;'>
                        <source src='data:{mime_type};base64,{audio_b64}' type='{mime_type}'>
                      </audio>
        
                      <div style='background:#18181b;padding:12px;border-radius:8px;font-size:12px;margin-top:10px;border:1px solid #27272a;'>
                        <div><b>Style:</b> <span style='color:#cbd5e1;'>{active_style[:140]}...</span></div>
                        <div style='display:flex;justify-content:space-between;margin-top:6px;'>
                          <span><b>Seed:</b> <code style='color:#38bdf8;'>{active_seed}</code></span>
                          <span><b>Sampler:</b> {sampler_name} / {scheduler}</span>
                          <span><b>Bitrate:</b> {audio_quality}</span>
                        </div>
                      </div>
                    </div>
                    """))
                else:
                    print("⚠️ No audio output found")
        
        except Exception as e:
            print(f"❌ Error: {e}")
            import traceback
            traceback.print_exc()

generate_button.on_click(on_generate_clicked)

# Display all widgets
display(HTML("<h2 style='color:#667eea;'>🎼 Style & Lyrics</h2>"))
display(genre_dropdown)
display(style_textarea)
display(lyrics_textarea)

display(HTML("<h2 style='color:#667eea;margin-top:20px;'>⚙️ Generation Settings</h2>"))
display(composition_radio)
display(abc_checkbox)
display(abc_tokens_dropdown)
display(sampler_dropdown)
display(scheduler_dropdown)
display(steps_slider)
display(randomize_checkbox)
display(seed_input)
display(format_radio)
display(quality_dropdown)

display(HTML("<br>"))
display(generate_button)
display(output_area)
```

```python
# @title 5. File Manager: Download All Generated Tracks
# @markdown Compresses all audio files in the output directory into a single `.zip` and provides a download link.
import glob
import os
import shutil
import time
from IPython.display import FileLink, HTML, display

output_dir = "/kaggle/working/ComfyUI/output"
# Check both root output and audio subfolder
all_audio = glob.glob(f"{output_dir}/*.mp3") + glob.glob(f"{output_dir}/*.wav") + glob.glob(f"{output_dir}/*.flac")
all_audio = list(set(all_audio)) # Remove duplicates if any

if not all_audio:
    display(HTML("<div style='background-color:#fff3cd;padding:15px;border-radius:8px;border-left:5px solid #ffc107;'><b style='color:#856404;'>⚠️ No audio tracks found.</b><br><span style='color:#856404;'>Generate some tracks first using Block 4.</span></div>"))
else:
    print(f"📁 Found {len(all_audio)} track(s):")
    print("-" * 70)
    
    total_size = 0
    for i, track in enumerate(sorted(all_audio, key=os.path.getctime, reverse=True), 1):
        size_mb = round(os.path.getsize(track) / (1024*1024), 2)
        total_size += size_mb
        filename = os.path.basename(track)
        created_time = time.strftime('%Y-%m-%d %H:%M:%S', time.localtime(os.path.getctime(track)))
        print(f"  {i}. {filename:<50} {size_mb:>6.2f} MB  [{created_time}]")
    
    print("-" * 70)
    print(f"📊 Total: {len(all_audio)} tracks, {round(total_size, 2)} MB")
    print()
    
    # Create zip archive
    zip_name = "/kaggle/working/YuE2_Generated_Tracks"
    print("📦 Creating zip archive...")
    shutil.make_archive(zip_name, 'zip', output_dir)
    
    zip_path = f"{zip_name}.zip"
    zip_size_mb = round(os.path.getsize(zip_path) / (1024*1024), 2)
    
    display(HTML(f"""
    <div style='background: linear-gradient(135deg, #667eea 0%, #764ba2 100%); color: white; padding: 25px; border-radius: 15px; text-align: center; box-shadow: 0 10px 30px rgba(0,0,0,0.3); margin: 20px 0;'>
        <h2 style='margin: 0 0 15px 0; font-size: 1.8em;'>📦 Download All Tracks</h2>
        <p style='font-size: 1.1em; margin-bottom: 20px;'>Archive contains <b>{len(all_audio)}</b> tracks ({zip_size_mb} MB)</p>
    </div>
    """))
    
    # Kaggle native download link
    display(FileLink(zip_path, result_html_prefix="<div style='text-align:center; margin: 10px 0; font-size: 1.2em;'><b>👇 Click here to download:</b><br>"))
```
